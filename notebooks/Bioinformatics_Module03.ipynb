####**TASK 12.3** <font color = orange> **Complete Rosalind Problem #3 Conditions and Loops** </font>

````python
# Enter code for Rosalind 3 here:
def sum_odd_integers(a, b):
    # If 'a' is even, shift to the first odd number in range
    if a % 2 == 0:
        a += 1
    # Step by 2 to sum only odd numbers through 'b' inclusively
    return sum(range(a, b + 1, 2))

# Example usage:
print(sum_odd_integers(4771, 9747))
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
def count_words_with_count(s):
    words = s.split()
    # Use set() to get unique words so we don't recount duplicates
    unique_words = set(words)

    for word in unique_words:
        # Use list.count() to count occurrences of 'word' in the list
        print(f"{word} {words.count(word)}")

# Example usage:
s = "We tried the list we tried the dict we tried the Whole shebang"
count_words_with_count(s)
````
